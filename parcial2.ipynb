### 1.1 `contar_rec(lista, x)` y `contar_it(lista, x)`

Escriba dos versiones de la función que retorna cuántas veces aparece el elemento `x` en la lista:

- `contar_rec(lista, x)`: versión **recursiva** (sin ciclos `for` ni `while`).
- `contar_it(lista, x)`: versión **iterativa** (con un ciclo `for` o `while`, sin recursión).

En ninguna de las dos versiones puede usar el método `.count()`.

```python
contar_rec([3, 1, 3, 2, 3, 5], 3)   # 3        contar_it([3, 1, 3, 2, 3, 5], 3)   # 3
contar_rec(["a", "b", "a"], "c")    # 0        contar_it(["a", "b", "a"], "c")    # 0
contar_rec([], 7)                   # 0        contar_it([], 7)                   # 0
```

````python
def contar_rec(lista, x):
    if len(lista) == 0:
        return 0
    primero = lista[0]
    resto = lista[1:]
    if primero == x:
        return 1 + contar_rec(resto, x)
    else:
        return 0 + contar_rec(resto, x)
````

````python
def contar_it(lista, x):
    contador = 0
    for i in range(len(lista)):
        if lista[i] == x:
            contador = contador + 1
    return contador
````

````python
# Test
for f in [contar_rec, contar_it]:
    assert f([3, 1, 3, 2, 3, 5], 3) == 3
    assert f(["a", "b", "a"], "c") == 0
    assert f(["a", "b", "a"], "a") == 2
    assert f([], 7) == 0
print("contar_rec y contar_it: OK")
````

### 1.2 Recursión: `suma_anidada(lista)`

Recibe una lista anidada de números (que puede contener otras listas) y retorna la suma de todos
los números, sin importar qué tan anidados estén. **No** puede usar ciclos `for` ni `while`, ni la
función `sum`.

```python
suma_anidada([1, [2, 3], [4, [5, [6]]], 7])
# 28
suma_anidada([[[]], 8])
# 8
```

Ayuda: si la lista es vacía debe retornar cero.

````python
def suma_anidada(lista):
    if len(lista) == 0:
        return 0
    primero = lista[0]
    resto = lista[1:]
    if type(primero) == list:
        suma_primero = suma_anidada(primero)
    else:
        suma_primero = primero
    suma_resto = suma_anidada(resto)
    return suma_primero + suma_resto
````

````python
# Test
assert suma_anidada([1, [2, 3], [4, [5, [6]]], 7]) == 28
assert suma_anidada([[[]], 8]) == 8
assert suma_anidada([]) == 0
print("suma_anidada: OK")
````

### 1.3 Recursión de árbol: `binarias(n)`

Escriba la función `binarias(n)` que retorne una lista con **todas** las cadenas de longitud `n`
formadas por `'0'` y `'1'` que **no** tienen dos `'1'` consecutivos.

```python
binarias(3)
# ['000', '001', '010', '100', '101']
len(binarias(10))
# 144
```

````python
def armar(cadena, n, resultado):
    if len(cadena) == n:
        resultado.append(cadena)
    else:
        armar(cadena + "0", n, resultado)
        if len(cadena) == 0:
            armar(cadena + "1", n, resultado)
        elif cadena[len(cadena) - 1] == "0":
            armar(cadena + "1", n, resultado)


def binarias(n):
    resultado = []
    armar("", n, resultado)
    return resultado
````

````python
# Test
assert sorted(binarias(3)) == ['000', '001', '010', '100', '101']
assert len(binarias(10)) == 144
assert all("11" not in c and len(c) == 10 for c in binarias(10))
print("binarias: OK")
````

## Ejercicio 3 (1 punto)

**Determinar qué se imprime** (sin ejecutar el código) y explicar qué calcula `misterio(n)` para
cualquier entero `n >= 0`.

```python
def misterio(n, nivel=0):
    sangria = "  " * nivel
    if n < 10:
        print(sangria + "base", n)
        return n
    print(sangria + "entra", n)
    r = misterio(n // 10, nivel + 1) + n % 10
    print(sangria + "sale", n, "->", r)
    return r

print(misterio(4827))
```

**Salida:**

```
entra 4827
  entra 482
    entra 48
      base 4
    sale 48 -> 12
  sale 482 -> 14
sale 4827 -> 21
21
```

**¿Qué calcula `misterio(n)`?**

Calcula la suma de los dígitos de `n`. Por ejemplo, para 4827 da 4 + 8 + 2 + 7 = 21.

En cada llamada le quita el último dígito al número con `n // 10` y baja un nivel, hasta que queda un número de un solo dígito, que es el caso base y se retorna tal cual. Después, al ir saliendo de las llamadas, a cada resultado le suma el dígito que había quitado (`n % 10`). Por eso primero se imprimen todos los "entra" y luego los "sale" en orden inverso.

### Verificación

Después de responder, ejecute la siguiente celda para comparar su respuesta.

````python
def misterio(n, nivel=0):
    sangria = "  " * nivel
    if n < 10:
        print(sangria + "base", n)
        return n
    print(sangria + "entra", n)
    r = misterio(n // 10, nivel + 1) + n % 10
    print(sangria + "sale", n, "->", r)
    return r

print(misterio(4827))
````
